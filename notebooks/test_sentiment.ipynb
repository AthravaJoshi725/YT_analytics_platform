```python
from transformers import pipeline
import os

# local model path (offline)
local_path = r"C:\Users\Admin\.cache\huggingface\hub\models--cardiffnlp--twitter-roberta-base-sentiment\snapshots"

# get the latest snapshot folder dynamically
folder = [os.path.join(local_path, f) for f in os.listdir(local_path)][0]

# load model
sentiment_model = pipeline("sentiment-analysis", model=folder)
print("✅ Model loaded offline successfully!")
```

```python
# label mapping for cardiffnlp/twitter-roberta-base-sentiment
label_map = {
    "LABEL_0": "negative",
    "LABEL_1": "neutral",
    "LABEL_2": "positive"
}

# test comments
comments = ["This is awesome!", "I hated the audio quality.", "You are walking on road"]

for c in comments:
    result = sentiment_model(c)[0]
    label = label_map.get(result['label'], result['label'])
    print(f"{c} → {label} ({result['score']:.2f})")
```
